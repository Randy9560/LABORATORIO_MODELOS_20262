```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def inspeccion_sujetos(h):
    tf_pac_1 = 4
    pase_pac_1 = 0.1
    t_pac_1 = np.arange(0, tf_pac_1+pase_pac_1, pase_pac_1)
    x_pac_1 = 5*np.sin(2*np.pi*0.5*t_pac_1)
    
    matriz1 = np.array([t_pac_1, x_pac_1])
    
    tf_pac_2 = 10
    pase_pac_2 = 0.2
    t_pac_2 = np.arange(0, tf_pac_2+pase_pac_2, pase_pac_2)
    x_pac_2 = 1*np.sin(2*np.pi*(1/4)*t_pac_2)
    
    matriz2 = np.array([t_pac_2, x_pac_2])
    
    plt.figure(figsize=(20, 6))
    plt.subplot(1, 2, 1)
    plt.stem(matriz1[0], matriz1[1], label='Paciente 1', linefmt='orange', markerfmt='D')
    plt.plot(matriz1[0], matriz1[1], label='Paciente 1', color='orange', linestyle='--', )
    plt.title('Paciente 1')
    plt.xlabel('Tiempo (s)')
    plt.ylabel('Señal')
    plt.grid()
    plt.legend()
    plt.subplot(1, 2, 2)
    plt.stem(matriz2[0], matriz2[1], label='Paciente 2', linefmt='blue', markerfmt='o')
    plt.plot(matriz2[0], matriz2[1], label='Paciente 2', color='blue')
    plt.title('Paciente 2')
    plt.xlabel('Tiempo (s)')
    plt.ylabel('Señal')
    plt.grid()
    plt.legend()
    
    
    B = np.mean((matriz1[1]-h)**2)
    
    info = 0
    for x in matriz2[1]:
        if x < h:
            info += 1
    if info%2 == 0:
        data = "Información del sujeto 2 par"
    else:
        data = "Información del sujeto 2 impar"
    return data
```

```python
a = inspeccion_sujetos(0.9)
    
```
