Crear la gráfica que se muestra en la Figura 1 mediante una función que contenga como parámetros de entrada el periodo de muestreo, el tiempo final, la amplitud y la frecuencia de la señal. La función debe guardar la imagen generada; para esto, consulte la función plt.savefig( ).

image.png

```python
# Importación de las librerías necesarias.
import numpy as np
# Se importa la librería de matplotlib, que permite graficar las señales.
import matplotlib.pyplot as plt
```

```python
# Se define la función "fun_sen", que recibe los argumentos Periodo de muestreo, Ampli
# Tiempo final y la frecuencia de la señal.
def fun_sen(Ts,A,tf,f):
  """Esta función genera una señal sinuidal"""
  # Se define el rango de tiempo para la señal senoidal.
  t = np.arange(0,tf + Ts, Ts)
  # Se define la señal senoidal.
  y = A*np.sin(2*np.pi*f*t)

  # Se grafica la señal entregada.
  plt.figure(figsize=(10,3))
  plt.plot(t,y,"-r")
  plt.stem(t,y,"b")
  plt.grid()
  plt.xlabel("Tiempo (s)")
  plt.ylabel("Amplitud")
  plt.title("Señal Punto 3")
  plt.show()
  plt.savefig("Señal 3.jpg")

# Se grafica la señal respectiva con los parámetros descritos en la imagen.
grafica = fun_sen(0.1,1,10,1/2)
```
