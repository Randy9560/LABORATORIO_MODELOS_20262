En Python se pueden manipular listas y matrices sin necesidad de una librería especial. Sin embargo, NumPy (Numerical Python) es una biblioteca libre ampliamente usada en ingeniería debido a la gran cantidad de beneficios que proporciona a la hora de manipular datos.

## 1.1. Generación de matrices y manipulación de elementos

**Ejemplo 1: Inicializar un vector**

```python
B = np.array([1, 5, 9, 4, 6, 7, 3, 8, 2])
print(B)
```

**Ejemplo 2: Inicializar una matriz bidimensional**

```python
A = np.array([[1, 5, 4], [2, 3, 6], [4, 8, 9]])
print(A)
```

**Ejemplo 3:** Acceder al tercer elemento de la segunda fila de la matriz A. Las matrices de Numpy empiezan su numeración en 0.

```python
element = A[1, 2]
print("Matrix at 2 row and 3 column = ", element)
```

**Ejemplo 4:** Calcular manualmente el producto escalar entre la primera y tercera columna de A.

```python
esc = A[0,0] * A[0,2] + A[1,0] * A[1,2] + A[2,0] * A[2,2]
print(esc)
```

**Ejemplo 5:** Acceder a la segunda columna y a la tercera fila de A.

```python
row_3 = A[2, :]
column_2 = A[:, 1]
print(row_3)
print(column_2)
```

## 1.2. Operaciones

**Ejemplo 1:** Elevar todos los elementos de la matriz A al cuadrado.

```python
A_squared = A ** 2
print(A_squared)
```

**Ejemplo 2:** Multiplicar todos los elementos de la matriz A por 1.6.

```python
new_A = A * 1.6
print(new_A)
```

**Ejemplo 3:** Crear una matriz B, (a) hacer el producto matricial y (b) hacer la multiplicación elemento a elemento.

```python
B = np.array([[2, 1, 3], [7, 1, 8], [6, 4, 5]])
# (a) producto de matrices
C = A @ B # O también: np.matmul(A, B)
print(C)
# (b) producto elemento a elemento
D = A * B
print(D)
```

**Ejemplo 4:** Encontrar las dimensiones de la matriz A y el número de elementos.

```python
dimension = A.shape
total_elements = A.size
print("Dimension = ", dimension)
print("Total elements = ", total_elements)
```

**Ejemplo 5:** Hallar la transpuesta de A.

```python
A_transpose = A.T
print(A_transpose)
```

- `np.zeros()` = matriz de ceros
- `np.ones()` = matriz de unos
- `np.empty()` = matriz cuyo contenido inicial es aleatorio
- `np.arange()` = matriz en un rango de elementos
- `np.linspace()` = matriz con valores espaciados linealmente en un intervalo especificado

**Actividad 1:** Hacer una matriz por cada una de las funciones mencionadas.

## 1.4. Representación de polinomios

**Ejemplo 1:** Ingresar el polinomio. Los polinomios se pueden expresar ingresando en la función los coeficientes de la expresión como un vector.

```python
p = np.poly1d([1, 4, 5, 0, 2])
print(p)
```

**Ejemplo 2:** Evaluar el polinomio en `x = 0.5`.

```python
p(0.5)
```

Los polinomios se pueden sumar, restar, multiplicar y dividir (devuelve el cociente y el resto).

**Ejemplo 3:** Multiplicar los polinomios `p1` y `p2`.

```python
p1 = np.poly1d([1, 2, 3])
p2 = np.poly1d([1, 0, 3])
p12 = p1*p2
print(p12)
```

**Ejemplo 4:** Dividir el polinomio `p` por el polinomio `s = x + 1`.

```python
s = np.poly1d([1, 1])
cociente, residuo = p / s
print(cociente)
print(residuo)
```

**Ejemplo 5:** Hallar las raíces del polinomio.

```python
roots = p.r
print(roots)
```
