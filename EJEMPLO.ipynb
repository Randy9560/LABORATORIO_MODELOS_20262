```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#1

def estudio (vector, umbral ):
  time= np.linspace (0,20,len(vector))

  p2= np.polyfit(time,vector,2)
  y2_fit=np.polyval(p2,time)
  plt.figure(figsize=(10,3))
  plt.scatter(time,vector, marker='o', color='black' , label="Datos")
  plt.plot(time, y2_fit , '--', color='r',label="Ajuste lineal")
  plt.grid()
  plt.title('Ajuste cuadratico')
  plt.xlabel("Tiempo")
  plt.ylabel("Y")
  plt.legend ()

  for i in range(len(vector)):
        if vector[i] > umbral:
          vector[i] = umbral

  return time


umbral=0.4
vector= [0.08,0.11,0.18,0.26,0.35,0.47,0.58,0.66,0.73,0.78,0.74,0.69,0.61,0.6,0.6]
estudio (vector, umbral)

```

```python
#2

def seno (periodoMuestreo, tiempoFinal, amplitud, frecuencia):
  time= np.arange(0, tiempoFinal + periodoMuestreo, periodoMuestreo)
  y= amplitud* np.sin (2* np.pi * frecuencia * time)

  for i in range(len(y)):
        if y[i] < 0:
          y[i] = 0

  return time, y

duracion         = 35    #seg
periodo_muestreo = 1/2  #seg
frecuencia       = 1/10   #Hz
amplitud         = 15    #unidades

t, y = seno (periodo_muestreo, duracion, amplitud, frecuencia)

plt.figure() #(figsize=(10,5))
plt.plot (t,y,marker='o') #grafica continua
plt.xlabel("Tiempo (s)")
plt.ylabel("Amplitud")
plt.title('Senal sinosoidal generada')
plt.grid()

plt.show()

```
