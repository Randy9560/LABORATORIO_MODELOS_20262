```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def analisis_flujo(periodo,amplitud):
    #Para ambos sujetos es ts y tf igual y el timepo
    ts=0.05
    tf=40
    time = np.arange(0,tf+ts,ts)
    X1=[]
    X2=[]

    #Punto 1
    for i in time:
        X1.append(amplitud*np.sin(2*np.pi*(1/periodo)*i))
    for i in time:
        X2.append((amplitud/2)*np.sin(2*np.pi*((1/periodo)*2)*i))
    #Punto 2
    plt.plot(time,X1,"--",color= "red",label="Señal sujeto 1",)
    plt.plot(time,X2,"+",color="green",label="Señal sujeto 2")
    plt.xlabel("Tiempo (s)")
    plt.ylabel("Amplitud")
    plt.title("Modelo de señal de flujo sanguíneo")
    plt.legend()
    plt.grid()
    plt.plot()

    #Punto 3
    X1=np.array(X1)
    X1_MAX=np.max(X1)
    X1_MIN=np.min(X1)
    norm = (X1-X1_MIN)/(X1_MAX-X1_MIN)
    #print(norm)

    #Punto 4
    valores_norm=[]
    for i in norm:
        if i < 0.1:
            valores_norm.append(1)
    print("La cantidad de números en nrom que son menores a 0.1 son",len(valores_norm))

    #Punto 5
    info=[norm,X2]
    info=np.array(info)
    return info
```

```python
Matriz_info= analisis_flujo(10,10)

print("La matriz info es: ", Matriz_info)
```
