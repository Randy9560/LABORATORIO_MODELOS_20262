```python
#Librerias 
import matplotlib.pyplot as plt
import numpy as np 
```

```python
#PUNTO 1

def signal_creation(Perido_s,Tiempo_total,F1,A1,F2,A2 ):

    v_tiempo = np.arange(0,Tiempo_total+Perido_s,Perido_s)

    señal1 = A1*np.sin(2*np.pi*F1*v_tiempo)

    señal2 = A2*np.sin(2*np.pi*F2*v_tiempo)

    Matriz = np.column_stack((v_tiempo,señal1,señal2))

    return Matriz
```

```python
#PUNTO 2

def signal_graph(matriz, Estilo1, Estilo2 ):

    plt.figure(figsize=(10, 5))
    plt.plot(matriz[:,0], matriz[:,1], label='Señal 1', color='b', linestyle= Estilo1)
    plt.plot(matriz[:,0], matriz[:,2], label='Señal 2', color='r', linestyle= Estilo2)
    plt.title("SEÑALES DE PRUEBA")
    plt.xlabel("Tiempo (s)")
    plt.ylabel("Fuerza (N)")
    plt.legend()
    plt.grid()
```

```python
#PUNTO 3

def material_analysis(matriz):
    señal1 = matriz[:,1]
    señal2 = matriz[:,2] 
    #RMS   
    n = len(señal1)

    sum1 = []
    sum2 = []

    for j in range(n):
        sum1.append(señal1[j]**2) 
        sum2.append(señal2[j]**2) 

    rms1 = np.sqrt((1/n)*np.sum(sum1))
    rms2 = np.sqrt((1/n)*np.sum(sum2))

    #CF

    Cf1 = (np.abs(np.max(señal1)))/rms1

    Cf2 = (np.abs(np.max(señal2)))/rms2
    
    if Cf1 > Cf2:
        salida = "Se recomienda usar la señal 1 para estimular el material porque presenta el mayor valor CF"
    elif Cf2 > Cf1:
        salida = "Se recomienda usar la señal 2 para estimular el material porque presenta el mayor valor CF"
    elif Cf1 == Cf2:
        salida = "Se recomienda usar cualquiera de las dos señales para estimular el material porque presenta el mismo valor CF"

    return salida
```

```python
#EJECUTADOR

m = signal_creation(0.1,7,0.5,10,0.8,5)
signal_graph(m,"--",":")
t = material_analysis(m)
print(t)
```
