## Capítulo 2. Representación gráfica

Matplotlib es una biblioteca de Python para crear visualizaciones estáticas, animadas e interactivas.

Instalar usando pip en la terminal: `pip install matplotlib`.

```python
import matplotlib.pyplot as plt
import numpy as np # Ver capítulo 1: Matrices
```

Matplotlib grafica los datos en una figura `figure` generando el área donde se presentará el gráfico. En este espacio se almacenan los títulos, el gráfico, la leyenda, la barra de colores y demás componentes de la gráfica. El comando para graficar señales continuas en dos dimensiones es `plot`, el cual considera dos parejas ordenadas, siendo x e y dos vectores de las mismas dimensiones.

**Ejemplo 1:** Graficar una señal senoidal de color rojo, con cuadrícula, ejes y título.

```python
time = np.arange(0, 15, 0.1) # Consultar cada uno de los parámetros de la función
y = np.sin(time)
fig = plt.figure()
plt.plot(time,y, color='r')
plt.title("Señal senoidal")
plt.xlabel("Tiempo (s)")
plt.ylabel("Amplitud")
plt.grid()
```

Es posible cambiar el color, el tipo de línea y marcador utilizando argumentos adicionales en el comando `plot`.

**Colores básicos:** `b` azul, `g` verde, `r` rojo, `c` cyan, `m` magenta, `y` amarillo, `k` negro, `w` blanco.

**Estilos de línea:** `-` o `solid` línea sólida; `--` o `dashed` línea discontinua; `-.` o `dashdot` línea con guiones y puntos; `:` o `dotted` línea punteada; `' '` o `none` sin línea.

**Marcadores:** `.` punto; `o` círculo; `v` triángulo hacia abajo; `^` triángulo hacia arriba; `*` estrella; `+` cruz; `s` cuadrado; `p` pentágono; `d` diamante.

**Ejemplo 2:** Graficar la función seno en tres rangos diferentes usando estilos de línea, colores y marcadores diferentes para diferenciarlas.

```python
time = np.arange(0, 8, 0.2)
y1 = np.sin(time)
y2 = np.sin(time - np.pi/2)
y3 = np.sin(time - np.pi)
fig = plt.figure(figsize=(10, 3)) # Modificar el tamaño de la figura
plt.plot(time, y1, label='sin(t)', color='b', linestyle='--', marker='o')
plt.plot(time, y2, label='sin(t - pi/2)', color='k', linestyle=':', marker='d')
plt.plot(time, y3, label='sin(t - pi)', color='#639833', linestyle='-', marker='X')
plt.title("Señales senoidales")
plt.xlabel("Tiempo (s)")
plt.ylabel("Amplitud")
plt.legend()
plt.grid()
```

**Ejemplo 3:** Graficar las señales del ejemplo anterior en una gradilla de 3 filas y 1 columna.

```python
plt.figure(figsize=(10, 5))
plt.subplot(3, 1, 1) # Consultar qué indica cada uno de los argumentos
plt.plot(time, y1, 'b--o')
plt.grid()
plt.ylabel("sin(t)")
plt.title("Señales senoidales en subplot")
plt.subplot(3, 1, 2)
plt.plot(time, y2, ':kd')
plt.grid()
plt.ylabel("sin(t - pi/2)")
plt.subplot(3, 1, 3)
plt.plot(time, y3, '-X', color='#639833')
plt.grid()
plt.ylabel("sin(t - pi)")
plt.xlabel("Tiempo (s)")
```
