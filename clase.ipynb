Sln por método ecuaciones

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate
import scipy.signal as signal
```

```python
def sistema(t,y,m,c):
    
    F = np.where(t>=5, 0.1, 0)
    x1, x2 = y
    d_x1 = x2
    d_x2 = (1/m)*F - (c/m)*x2
    return [d_x1, d_x2]
```

```python
t_s = (0, 15)
y0 = [0, -1]
t_e = np.arange(0, 15 + 0.1, 0.1)
m = 1
c = 0.8

sol = integrate.solve_ivp(sistema, t_s, y0, method='RK45', t_eval=t_e, args=(m,c))

```

```python
plt.figure()
plt.plot(sol.t, sol.y[0])
plt.xlabel("tiempo")
plt.ylabel("Posición")

plt.figure()
plt.plot(sol.t, sol.y[1])
plt.xlabel("tiempo")
plt.ylabel("velocidad")
```

Sln por fn transferencia

```python
num = [1]
den = [m, c, 0]
sys = signal.TransferFunction(num, den)

F = np.where(t_e >= 5, 0.1, 0)


t_out, y_out, _ = signal.lsim(sys, F, T=t_e)


```

```python
plt.figure()
plt.plot(t_out, y_out)
plt.ylabel("desplazamiento")
```
