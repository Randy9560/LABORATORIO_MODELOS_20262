```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def inspeccion_sujetos (h):
    #PUNTO 1
    ts=0.1
    tf=4
    #Vector de tiempo sujeto 1
    tiempo = np.arange (0,tf+ts,ts)
    #Señal sujeto 1
    sujeto1 = np.sin(2 * np.pi * 1/2 * tiempo)
    #Matriz sujeto 1
    matriz1 = np.column_stack((tiempo, sujeto1))

    ts2=0.2
    tf2=10
    #Vector de tiempo sujeto 2
    tiempo2 = np.arange (0,tf2+ts2,ts2)
    #Señal sujeto 2
    sujeto2 = np.sin(2 * np.pi * 1/2 * tiempo2)
    #Matriz sujeto 1
    matriz2 = np.column_stack((tiempo2, sujeto2))

    #PUNTO 2
    plt.subplot(1,2,1)
    # -------- Grafica 1 --------
    plt.plot(tiempo,sujeto1,'r',linewidth = 2, linestyle= ':')
    plt.title("Sujeto 1")
    plt.xlabel("Tiempo (s)")
    plt.ylabel("Amplitud")
    plt.grid()

    # -------- Grafica 2 --------
    plt.subplot(1,2,2)

    plt.plot(tiempo2,sujeto2,'g',linewidth = 2, linestyle ='-')
    plt.title("Sujeto 2")
    plt.xlabel("Tiempo (s)")
    plt.ylabel("Amplitud")
    plt.grid()

    plt.tight_layout()
    plt.show()

    #PUNTO 3
    lista = []

    for i in matriz1[:,1]:
        x = (i - h)**2
        lista.append(x)

    z = sum(lista)
    B = z/len(lista)

    #PUNTO 4
    num=[]
    for c in sujeto2:
        if c < h:
            num.append(c)

    info = len(num)

    #PUNTO 5
    if info % 2 == 0:
        data = 'Información del sujeto 2 par'

    else:
        data = 'Información del sujeto 2 impar'  
    return data
```

```python
inspeccion_sujetos(5)
```
