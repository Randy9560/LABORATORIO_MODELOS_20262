### 4.1. Estructuras de flujo

Las estructuras de control en Python permiten dirigir el flujo de ejecución de un programa.

Estas estructuras son esenciales para tomar decisiones en el código, repetir operaciones y controlar el proceso de ejecución en función de distintas condiciones. Python ofrece varias estructuras de control, cada una adaptada a necesidades específicas.

A diferencia de otros lenguajes, Python utiliza los espacios en blanco (indentaciones) para definir la estructura del código. En los ejemplos que se muestran a continuación se resalta la importancia de tener código indentado para determinar dónde inicia y termina cada bloque de código.

### (A) Instrucción `if`

```python
edad = 37

if edad < 18:
    print("Menor de edad")
elif edad >= 18 and edad < 65:
    print("Adulto")
else:
    print("Adulto mayor")
```

**Salida:** `Adulto`

### (B) Instrucción `switch`

```python
nota = "B"

match nota:
    case "A":
        print("Excelente")
    case "B":
        print("Bueno")
    case "C":
        print("Regular")
    case "D":
        print("Malo")
    case _:  # Caso por defecto
        print("Nota no válida")
```

**Salida:** `Bueno`

### (C) Bucle `for`

```python
frutas = ["manzana", "banana", "cereza", "mango"]

for f in frutas:
    print("La fruta es", f)
```

**Salida:**
- La fruta es manzana
- La fruta es banana
- La fruta es cereza
- La fruta es mango

### (D) Bucle `while`

```python
contador = 0

while contador < 4:
    print("Dentro del bucle. Intento", contador + 1)
    contador += 1
```

**Salida:**
- Dentro del bucle. Intento 1
- Dentro del bucle. Intento 2
- Dentro del bucle. Intento 3
- Dentro del bucle. Intento 4

## 4.1. Funciones

Para definir una función en Python, se utiliza la palabra clave `def`, seguida del nombre de la función y en paréntesis los argumentos necesarios para su ejecución. El bloque de código indentado formará el cuerpo de la función con las actividades a realizar.

### Ejemplo 1

Crear una función que genere una matriz aleatoria de `n x n` con ceros en las esquinas.

```python
import numpy as np
import matplotlib.pyplot as plt

# Definición de la función
def matrizN(n):
    """Función que genera una matriz de nxn con valores aleatorios
    y sus esquinas en 0."""
    m = np.random.rand(n, n)
    m[0, 0] = 0
    m[0, -1] = 0
    m[-1, 0] = 0
    m[-1, -1] = 0
    return m

# Se llama a la función para poder utilizarla.
matriz = matrizN(5)
print(matriz)
```

### Ejemplo 2

Diseñar una función que genere un tren de impulsos rampa, cuyos argumentos de entrada sean el número de trenes, la duración de los pulsos y el periodo de muestreo.

```python
def tren(Tm, Np, T):
    """Función tren: retorna una señal digital con periodo de muestreo Tm,
    número de pulsos Np y duración T. Además, genera el vector de tiempo t
    y realiza la gráfica de la señal."""

    tb = np.arange(0, T + Tm, Tm)  # Genera la rampa
    n = len(tb)  # Se obtiene el número de elementos de la rampa

    r1 = np.concatenate((np.zeros(n), tb))  # Se crea un solo pulso

    # Se crea un vector para almacenar la señal con el número de pulsos ingresados
    r = np.array([])

    for i in range(Np):  # Creación de la señal con Np pulsos
        r = np.concatenate((r, r1))

    # Creación del vector de tiempo completo
    t = np.linspace(0, 2 * Np * T, 2 * Np * n)

    plt.figure()
    plt.stem(t, r)

    return r, t

# Ejecutar la función tren
r, t = tren(1, 5, 5)
```
