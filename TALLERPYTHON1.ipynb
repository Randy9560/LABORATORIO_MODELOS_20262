```python
X = [2,6,8,9,0,1,8]
Y = [0,1,3,6,4,2,7]
xmax= np.max(X)
ymax= np.max(Y)
print(f"Valor Maximo X= {xmax}\nValor Maximo Y= {ymax}")
```

```python
N=[]
for i in range(len(X)): # Iterate over indices instead of values
  if X[i] == Y[i]:
    N.append(0)
  elif X[i] > Y[i]:
    N.append(1)
  elif X[i] < Y[i]:
    N.append(-1)
print(N)
```

```python
for i in range(len((X))):
  if X[i] ==0:
    X[i]=i
for j in range(len((Y))):
  if Y[j] ==0:
    Y[j]=j
print(X)
print(Y)
```

```python
def fun_sin(Ts,A,tf,f):
  t=np.arange(0, tf+Ts, Ts)
  y=A*np.sin(2*np.pi*f*t)
  plt.figure(figsize=(10,3))
  plt.plot(t,y,"--r")
  plt.stem(t,y,"b")
  plt.grid()
  plt.xlabel("Tiempo(s)")
  plt.ylabel("Amplitud")
  plt.title("Señal punto 3")
  plt.show()


fun_sin(0.1,1,10,1/2)
```

```python
t1=np.arange(0,5+0.1, 0.1)
y1=1*np.sin(2*np.pi*1*t1)
t2=np.arange(0,0.5+(0.2/20),(0.2/20))
y2=10*np.sin(2*np.pi*1/0.2*t2)

plt.figure(figsize=(10,8))
plt.subplot(2,1,1)
plt.title("Comparación de Seno")
plt.plot(t1,y1,"--r")
plt.stem(t1,y1,"b")
plt.grid()
plt.xlabel("Tiempo(s)")
plt.ylabel("Amplitud")
plt.subplot(2,1,2)
plt.plot(t2,y2,"m")
plt.stem(t2,y2,"g")
plt.xlabel("Tiempo(s)")
plt.ylabel("Amplitud")
plt.grid()
plt.show()
```

```python
t6=np.arange(0,10+0.1, 0.1)
y6=2*np.sin(2*np.pi*1/4*t6)
Matriz6=np.array([t6,y6]).T
def datos6( matriz):

  tiempo = matriz[:, 0]
  amplitud = matriz[:, 1]

  plt.plot(tiempo, amplitud, "--r")
  plt.xlabel("Tiempo")
  plt.ylabel("Amplitud")
  plt.title("A) Amplitud vs Tiempo")
  plt.grid()
  plt.show()

  promedio = np.mean(amplitud)
  print("B) El promedio de la amplitud es:", promedio)

datos6(Matriz6)
```

```python
X8=np.array([0,2,5,7,15])
Y8=np.array([3,7,10,15,26])

m,b= np.polyfit(X8, Y8, 1)
Y_recta=m* X8+b
plt.figure(figsize=(10, 10))
plt.scatter(X8, Y8, color="g", label="Datos")
plt.plot(X8, Y_recta, color="r", label="Recta de ajuste")
plt.grid()
plt.xlabel("Vector X")
plt.ylabel("Vector Y")
plt.title("Ajuste lineal")
plt.legend()
plt.show()

# Mostrar ecuación
print("Pendiente:", round(m,2))
print("Intercepto:", round(b,2))
print("Ecuación: Y =", round(m,2), "* X +",  round(b,2))
```

```python
def curvaty(tm,f,A,tetha):
  t=np.arange(-4,5,tm)
  y= (5*(np.exp(-2*t)))+(A*np.sin((2*np.pi*f*t)+tetha))
  plt.figure(figsize=(10,8))
  plt.plot(t,y,"r",marker='X')
  plt.title("Función Curvaty")
  plt.xlabel("Tiempo(s)")
  plt.ylabel("Amplitud")
  plt.grid()
  plt.show()

curvaty(0.1,1/2,1,0)
```

```python
a1=np.random.randint(1,101,1000)
a2=np.random.randint(1,101,1000)
```

```python
corr=np.corrcoef(a1,a2)
print(f"La matriz de correlación es:\n \n{corr}")
```

```python
aj=np.polyfit(a1,a2,2)
aj2=np.polyval(aj,a1)
plt.figure(figsize=(10,10))
plt.scatter(a1,a2,color="m", label="vectores")
plt.plot(a1,aj2,"b", label="Ajuste Lineal")
plt.grid()
plt.legend()
plt.xlabel("Vector 1")
plt.ylabel("Vector 2")
plt.show()
```

```python
vect=[]
for i in range(1000):
  er=a1[i]-a2[i]
  vect.append(er)
vect=np.array(vect)
J=1/1000* np.sum(vect)
print(f"El ERROR es: {J}")
```
